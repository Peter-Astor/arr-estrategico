# 03_robustez_y_suplementarios.ipynb
Análisis de robustez y modelos extendidos

```python
import pandas as pd
import statsmodels.api as sm
import statsmodels.formula.api as smf

df = pd.read_csv("dataset_between.csv")
df.head()
```

## Modelo con covariables demográficas

```python
model_demo = smf.gee(
"""Mantiene_bin ~ C(Tratamiento)*Expectativa_Activa
+ NDC_c
+ SDO_c
+ politica
+ nivel_se
+ Genero""",groups="ID_Sujeto",
data=df,
family=sm.families.Binomial(),
cov_struct=sm.cov_struct.Exchangeable()
).fit()

print(model_demo.summary())
```

## Modelo con moderación por Gap

```python
model_gap = smf.gee(
"""Mantiene_bin ~ C(Tratamiento)*Expectativa_Activa
+ Gap0
+ Gap_pos
+ Gap0*Expectativa_Activa
+ Gap_pos*Expectativa_Activa""",groups="ID_Sujeto",
data=df,
family=sm.families.Binomial(),
cov_struct=sm.cov_struct.Exchangeable()
).fit()

print(model_gap.summary())
```

## Subgrupo alta Necesidad de Cognición

```python
df_high = df[df["NDC_Score"] >= df["NDC_Score"].quantile(.75)]

model_high = smf.gee(
"Mantiene_bin ~ C(Tratamiento)*Expectativa_Activa",
groups="ID_Sujeto",
data=df_high,
family=sm.families.Binomial(),
cov_struct=sm.cov_struct.Exchangeable()
).fit()

print(model_high.summary())
```
