# 01_preprocesamiento.ipynb
Preparación del dataset analítico (between-subject)

```python
import pandas as pd
import numpy as np

df = pd.read_csv("df_long0.csv")
print(df.shape)
df.head()
```

## Filtrar primer bloque (análisis between-subject)

```python
df_between = df[df["Orden_1"] == 1].copy()
print(df_between.shape)
```

## Crear variables analíticas

```python
# Tratamiento
df_between["Tratamiento"] = df_between["Bloque"]
df_between["Tratamiento"] = pd.Categorical(
    df_between["Tratamiento"],
    categories=["DIST","SIN","CON"]
)

# Variable dependiente
df_between["Mantiene_bin"] = df_between["Mantiene"].astype(int)

# Centrar NDC
df_between["NDC_c"] = df_between["NDC_Score"] - df_between["NDC_Score"].mean()

# Centrar SDO
df_between["SDO_c"] = df_between["SDO_Score"] - df_between["SDO_Score"].mean()

# Gap
df_between["Gap"] = df_between["Gap_Size"]

# Dummy Gap = 0
df_between["Gap0"] = (df_between["Gap"] == 0).astype(int)

# Gap positivo centrado
gap_pos_mean = df_between.loc[df_between["Gap"]>0, "Gap"].mean()
df_between["Gap_pos"] = df_between["Gap"].where(df_between["Gap"]>0)
df_between["Gap_pos"] = df_between["Gap_pos"] - gap_pos_mean

df_between.head()
```

## Guardar dataset analítico

```python
df_between.to_csv("dataset_between.csv", index=False)
print("Dataset guardado como dataset_between.csv")
```
