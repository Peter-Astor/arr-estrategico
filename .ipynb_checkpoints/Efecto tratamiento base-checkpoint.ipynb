```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df_long = pd.read_csv('df_long.csv')
df_delta=pd.read_csv('df_delta.csv')
df_resumen = pd.read_csv('Base_Resumen_Sujetos.csv')
```

```python
# A. Preparación para Gráfico 1 (Uso de Base Resumen)
df_t_long = df_resumen.melt(
    id_vars='ID_Sujeto', 
    value_vars=['Promedio_CON', 'Promedio_SIN', 'Promedio_DIST'],
    var_name='Tratamiento', 
    value_name='Promedio'
)
df_t_long['Tratamiento'] = df_t_long['Tratamiento'].str.replace('Promedio_', '')

# B. Preparación para Gráfico 2 (Primer Escenario)
df_primero = df_long.drop_duplicates(subset=['ID_Sujeto'], keep='first').copy()
# Extracción robusta (insensible a mayúsculas/minúsculas)
df_primero['Tratamiento'] = df_primero['Dilema'].str.upper().str.extract('(CON|SIN|DIST)', expand=False)

orden = ['CON', 'SIN', 'DIST']

# --- 2. VISUALIZACIÓN INTEGRADA ---
sns.set_theme(style="whitegrid")
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

# Gráfico 1: Comportamiento General
sns.barplot(data=df_t_long, x='Tratamiento', y='Promedio', order=orden,
            errorbar='se', capsize=.1, palette='Set2', ax=ax1)
ax1.set_title('Promedio General por Tratamiento\n(Comportamiento Sostenido)', fontsize=13, fontweight='bold')

# Gráfico 2: Reacción "Virgen" (Primer Escenario)
sns.barplot(data=df_primero, x='Tratamiento', y='Mantiene', order=orden,
            errorbar='se', capsize=.1, palette='Set2', ax=ax2)
sns.stripplot(data=df_primero, x='Tratamiento', y='Mantiene', order=orden,
              color='black', alpha=0.3, jitter=True, ax=ax2)
ax2.set_title('Primer Escenario Encontrado\n(Tasa de Mantiene - Reacción Inicial)', fontsize=13, fontweight='bold')

# Ajustes comunes
for ax in [ax1, ax2]:
    ax.set(ylabel='Proporción / Tasa Media (± SEM)', xlabel='Tratamiento', ylim=(0, 1.1))

plt.tight_layout()
plt.savefig('panel_comparativo_final.png', dpi=300)
plt.show()

# --- 3. TABLAS RESUMEN (Uso de .reindex para evitar el KeyError) ---
print("--- TABLA 1: PROMEDIOS GENERALES (SOSTENIDO) ---")
tabla_1 = df_t_long.groupby('Tratamiento')['Promedio'].agg(['count', 'mean', 'sem']).reindex(orden)
print(tabla_1.round(3))

print("\n--- TABLA 2: REACCIÓN INICIAL (PRIMER ESCENARIO) ---")
tabla_2 = df_primero.groupby('Tratamiento')['Mantiene'].agg(['count', 'mean', 'sem']).reindex(orden)
print(tabla_2.round(3))
```
